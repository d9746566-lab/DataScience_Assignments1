```python
1.What is Normalization & Standardization and how is it helpful?
Normalization
Normalization is the process of rescaling the values of a feature to a common range, usually [0, 1].
Formula :    
         Xnorm = X - Xmin / Xmax - Xmin
Example:
If a feature has values between 10 and 100, normalization will transform it to values between 0 and 1.
When to Use:
When using distance-based algorithms like:   
             K-Nearest Neighbors (KNN)
     K-Means clustering
When features are on different scales and need to be brought to the same scale.
Standardization
Standardization
Standardization (or Z-score normalization) transforms the data to have a mean of 0 and a standard deviation of 1.
Formula :       
        Xstd = X−μ / σ
​Where:
μ is the mean
σ is the standard deviation
When to Use:
When your data has outliers.
For linear models (like regression), SVM, or PCA, where assumptions of normally distributed features are important.
2.What techniques can be used to address multicollinearity in multiple linear regression?
 ##### What is Multicollinearity

Multicollinearity occurs when two or more independent variables in a regression model are highly correlated.
This can make it difficult to determine the individual effect of each variable, leading to unreliable coefficient estimates and inflated standard errors.

##### Removing Correlated Variables

One of the simplest ways to handle multicollinearity is to remove one of the correlated variables, especially if it does not significantly improve the model's performance or interpretability.

##### Principal Component Analysis (PCA)

PCA is a dimensionality reduction technique that transforms correlated variables into a new set of uncorrelated components. 
This helps in preserving the information while eliminating multicollinearity.   
##### Regularization Techniques

Ridge Regression (L2) and Lasso Regression (L1) are powerful methods that address multicollinearity by penalizing large coefficients:
Ridge shrinks all coefficients but keeps them in the model.
Lasso can reduce some coefficients to zero, effectively performing feature selection.

##### Variance Inflation Factor (VIF)

VIF measures how much the variance of a regression coefficient is inflated due to multicollinearity.
A VIF above 5 or 10 is typically a warning sign. Variables with high VIFs may need to be removed or transformed.

##### Using Domain Knowledge

Sometimes, multicollinearity arises because multiple features represent similar concepts. 
In such cases, domain knowledge can guide which variable is more meaningful and should be retained in the model.
```
