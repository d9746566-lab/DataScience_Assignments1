# Exercise 1:Prime Numbers

```python
# Program to check whether a number is Prime or Not

# Taking input from the user
num = int(input("Enter a number: "))

# Checking numbers less than or equal to 1
if num <= 1:
    print(num, "is not a Prime Number")

else:
    # Loop starts from 2 and goes till square root of the number
    for i in range(2, int(num**0.5) + 1):

        # Check whether the number is divisible
        if num % i == 0:
            print(num, "is not a Prime Number")
            break

    else:
        print(num, "is a Prime Number")
```

# Exercise 2:Product of Random Numbers

```python
# Importing random module
import random

# Generating two random numbers between 1 and 10
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

# Displaying the random numbers
print("First Number :", num1)
print("Second Number :", num2)

# Taking user's answer
user_answer = int(input("Enter the product of these numbers: "))

# Calculating the correct product
correct_answer = num1 * num2

# Checking whether the answer is correct
if user_answer == correct_answer:
    print("Correct Answer!")
else:
    print("Wrong Answer!")
    print("Correct Product is:", correct_answer)
```

# Exercise 3:Squares of Even/Odd Numbers

```python
# Program to print squares of even numbers from 100 to 200

# Chosen: Even Numbers

print("Squares of Even Numbers from 100 to 200")

# Loop from 100 to 200
for i in range(100, 201):

    # Check whether the number is even
    if i % 2 == 0:

        # Print the square of the even number
        print(i, "=", i ** 2)
```

# Exercise 4:Word counter

```python
# Function to count the number of words in a string

def count_words(text):
    # Split the string into words
    words = text.split()

    # Return the number of words
    return len(words)

# Taking input from the user
sentence = input("Enter a sentence: ")

# Calling the function
result = count_words(sentence)

# Displaying the result
print("Number of words:", result)

# Testing with another string
test = "Python is easy to learn"
print("Test String:", test)
print("Number of words:", count_words(test))
```

# Exercise 5:Check for Palindrome

```python
# Function to check whether a string is a palindrome

def check_palindrome(text):

    # Convert to lowercase
    text = text.lower()

    # Remove spaces
    text = text.replace(" ", "")

    # Check palindrome
    if text == text[::-1]:
        return "Palindrome"
    else:
        return "Not a Palindrome"

# Taking input from the user
sentence = input("Enter a string: ")

# Calling the function
result = check_palindrome(sentence)

# Displaying the result
print(result)

# Testing with another string
test = "Madam"
print("Test String:", test)
print(check_palindrome(test))
```
